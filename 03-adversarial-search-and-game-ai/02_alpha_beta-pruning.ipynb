<h1> Alpha-Beta Pruning Algorithm </h1>

Alpha-Beta says: </br>
If a part of the tree has no other impact on the final decision, don't examine it.

That is:</br>
Pruning = removing unnecessary branches

**Alpha (α):** </br>
The best guaranteed value for MAX so far.</br>
That is: The minimum value MAX can take is this.

**Beta (β):**</br>
The best guaranteed value for MIN.</br>
That is: The maximum value MIN allows is this.

Alpha-Beta = Minimax + Pruning
