### Minimax Tie Between Multiple Optimal Actions

In some states, Minimax may return an action that is not the most intuitive move, even though the algorithm is working correctly.

The reason is that multiple actions can lead to the same final utility value. For example, in a Tic-Tac-Toe state:

- Action 5 may create an immediate win.
- Action 2 may also guarantee a win after considering the opponent's optimal responses.

Since the current implementation updates the best action only when:

`value > best_value`

the first action with the maximum value is selected. Therefore, if multiple actions have the same utility, the earliest action in the action list is returned.

This behavior is not an error; it is a tie-breaking issue. In practical game agents, additional strategies such as random selection among equally optimal actions or heuristic-based tie-breaking can be used.

```python
game = TicTacToe()

state = game.initial_state()
print(state)

action = depth_limited_minimax_decision(
    game,
    state,
    depth_limit=3,
    evaluation_function=evaluate_tic_tac_toe,
)

action
```

**Important point**

In Tic-Tac-Toe, the center is usually the best starting move, as it creates the most potential winning lines.


Full Minimax may initially see every move as equal in terms of final outcome, as the full game is played all optimal moves can produce a Draw. </br>
But Depth-Limited, because it works with Evaluation, sees the differences in heuristic value and chooses the center.

Depth-Limited is dependent on Evaluation Function Quality and if Evaluation is poor, it may choose a wrong move.

### Depth-Limited Minimax Behavior

Depth-Limited Minimax reduces computational cost by stopping the search at a predefined depth and using an evaluation function instead of the exact utility.

Increasing depth generally increases the number of explored nodes. However, if the game reaches terminal states before the depth limit, increasing the limit further does not increase the search space.

Deeper search may also change the selected action because the agent can discover additional optimal paths. When multiple actions have the same final utility, the selected action depends on the tie-breaking strategy.
