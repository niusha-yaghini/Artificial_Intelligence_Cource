<h1> DFS </h1>

```python
result = dfs(
    graph=graph,
    start="A",
    goal="H",
)

print("Success:", result.success)
print("Path:", result.path)
print("Traversal:", result.traversal_order)
print("Expanded:", result.nodes_expanded)
print("Generated:", result.nodes_generated)
print("Max frontier:", result.max_frontier_size)
print("Solution depth:", result.solution_depth)
```

Comparison

```python
bfs_result = bfs(graph, "A", "H")
dfs_result = dfs(graph, "A", "H")


print("BFS")
print("Traversal:", bfs_result.traversal_order)
print("Path:", bfs_result.path)
print("Solution depth:", bfs_result.solution_depth)
print("Expanded:", bfs_result.nodes_expanded)
print()

print("DFS")
print("Traversal:", dfs_result.traversal_order)
print("Path:", dfs_result.path)
print("Solution depth:", dfs_result.solution_depth)
print("Expanded:", dfs_result.nodes_expanded)
```

The performance of the Search Algorithm depends heavily on the structure of the State Space.

```python
deep_graph = Graph()

deep_graph.add_edge("A", "B")
deep_graph.add_edge("A", "C")

deep_graph.add_edge("B", "D")
deep_graph.add_edge("D", "E")
deep_graph.add_edge("E", "F")
deep_graph.add_edge("F", "X")

deep_graph.add_edge("C", "G")
```

```python
bfs_result = bfs(
    deep_graph,
    "A",
    "G",
)

dfs_result = dfs(
    deep_graph,
    "A",
    "G",
)
```

```python
print("BFS Traversal:")
print(bfs_result.traversal_order)

print()

print("DFS Traversal:")
print(dfs_result.traversal_order)
```

BFS:
memory expensive

DFS:
memory efficient

Visited list is even more critical for DFS than BFS.

If we search on a Tree, we usually don't have a Cycle.

But on a Graph we may have: </br>
Repeated State
Cycle
Multiple Paths

Tree Search: frontier is enough

Graph Seasrch: frontier + visited

<h1> Depth-Limited Search </h1>

```python
result = depth_limited_search(
    graph,
    start="A",
    goal="H",
    depth_limit=2,
)

print(result.success)
print(result.path)
print(result.traversal_order)
```

```python
result = depth_limited_search(
    graph,
    start="A",
    goal="H",
    depth_limit=3,
)

print(result.success)
print(result.path)
```
