Unreachable Goal Test

```python
disconnected_graph = Graph()

disconnected_graph.add_edge("A", "B")
disconnected_graph.add_edge("B", "C")

disconnected_graph.add_node("Z")
```

A --- B --- C

```python
result = bfs(
    disconnected_graph,
    "A",
    "Z",
)

print(result.success)
print(result.path)
print(result.traversal_order)
```

This means that the Search Space is exhausted but the Goal is not found.
