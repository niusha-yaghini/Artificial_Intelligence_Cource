```python
from src.grid import GridEnvironment

from src.search import (
    bfs,
    greedy_best_first_search,
    a_star_search,
    uniform_cost_search,
)

from src.heuristics import (
    manhattan_distance,
)

from src.visualization import print_grid
```

```python
grid = GridEnvironment(
    rows=6,
    cols=8,
    obstacles={
        (0, 3),
        (1, 1),
        (1, 3),
        (1, 5),
        (2, 1),
        (2, 5),
        (3, 2),
        (3, 3),
        (4, 0),
        (4, 5),
        (5, 2),
    },
)

start = (0, 0)
goal = (5, 7)
```

```python
bfs_result = bfs(
    grid,
    start,
    goal,
)

print("Success:", bfs_result.success)
print("Path:", bfs_result.path)
print("Depth:", bfs_result.solution_depth)
print("Cost:", bfs_result.path_cost)
print("Expanded:", bfs_result.nodes_expanded)
```

```python
greedy_result = greedy_best_first_search(
    grid,
    start,
    goal,
    manhattan_distance,
)

astar_result = a_star_search(
    grid,
    start,
    goal,
    manhattan_distance,
)
```

```python
print("Greedy")
print("Path length:", greedy_result.solution_depth)
print("Expanded:", greedy_result.nodes_expanded)

print()

print("A*")
print("Path length:", astar_result.solution_depth)
print("Expanded:", astar_result.nodes_expanded)
```

```python
print_grid(
    grid,
    start=start,
    goal=goal,
    path=astar_result.path,
)
```

```python
print_grid(
    grid,
    start=start,
    goal=goal,
    path=bfs_result.path,
    explored=bfs_result.traversal_order,
)
```

```python
print_grid(
    grid,
    start=start,
    goal=goal,
    path=astar_result.path,
    explored=astar_result.traversal_order,
)
```

<h1> Weighted Terrain </h1>

```python
weighted_grid = GridEnvironment(
    rows=5,
    cols=7,
    obstacles={
        (1, 3),
        (2, 3),
        (3, 3),
    },
    terrain_costs={
        (0, 2): 7,
        (0, 3): 7,
        (0, 4): 7,
        (1, 2): 5,
        (2, 2): 5,
    },
)

start = (0, 0)
goal = (0, 6)
```

```python
bfs_result = bfs(
    weighted_grid,
    start,
    goal,
)

ucs_result = uniform_cost_search(
    weighted_grid,
    start,
    goal,
)

print("BFS")
print("Path:", bfs_result.path)
print("Depth:", bfs_result.solution_depth)
print("Cost:", bfs_result.path_cost)

print()

print("UCS")
print("Path:", ucs_result.path)
print("Depth:", ucs_result.solution_depth)
print("Cost:", ucs_result.path_cost)
```

for good heuristic:


h(n)=distance(n,goal)×min_step_cost

```python
bfs_result = bfs(
    weighted_grid,
    start,
    goal,
)

ucs_result = uniform_cost_search(
    weighted_grid,
    start,
    goal,
)

greedy_result = greedy_best_first_search(
    weighted_grid,
    start,
    goal,
    manhattan_distance,
)

astar_result = a_star_search(
    weighted_grid,
    start,
    goal,
    manhattan_distance,
)
```

```python
results = {
    "BFS": bfs_result,
    "UCS": ucs_result,
    "Greedy": greedy_result,
    "A*": astar_result,
}

for name, result in results.items():
    print(name)
    print("  Depth:", result.solution_depth)
    print("  Cost:", result.path_cost)
    print("  Expanded:", result.nodes_expanded)
    print("  Generated:", result.nodes_generated)
    print("  Max Frontier:", result.max_frontier_size)
    print()
```
