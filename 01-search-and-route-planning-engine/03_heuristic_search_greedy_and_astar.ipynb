```python
current = (3, 4)
goal = (9, 10)

print(
    "Manhattan:",
    manhattan_distance(current, goal),
)

print(
    "Euclidean:",
    euclidean_distance(current, goal),
)

print(
    "Chebyshev:",
    chebyshev_distance(current, goal),
)
```

<h1> Greedy </h1>

A geometric graph for testing

```python
graph = Graph(directed=False)

graph.add_edge((0, 0), (1, 0), 1)
graph.add_edge((1, 0), (2, 0), 1)
graph.add_edge((2, 0), (3, 0), 1)

graph.add_edge((0, 0), (0, 1), 1)
graph.add_edge((0, 1), (1, 1), 1)
graph.add_edge((1, 1), (2, 1), 1)
graph.add_edge((2, 1), (3, 1), 1)

graph.add_edge((3, 1), (3, 0), 1)
```

```python
start = (0, 0)
goal = (3, 0)

greedy_result = greedy_best_first_search(
    graph=graph,
    start=start,
    goal=goal,
    heuristic=manhattan_distance,
)

print("Path:", greedy_result.path)
print("Cost:", greedy_result.path_cost)
print("Traversal:", greedy_result.traversal_order)
print("Expanded:", greedy_result.nodes_expanded)
```

Greedy might go that way because of A's proximity to Goal, even if the next Edge is very expensive.

<h1> A* </h1>

g(n)
= actual cost so far

h(n)
= estimated remaining cost

f(n)
= estimated total solution cost

```python
astar_result = a_star_search(
    graph=graph,
    start=start,
    goal=goal,
    heuristic=manhattan_distance,
)

print("Path:", astar_result.path)
print("Cost:", astar_result.path_cost)
print("Traversal:", astar_result.traversal_order)
print("Expanded:", astar_result.nodes_expanded)
```

```python
ucs_result = uniform_cost_search(
    graph,
    start,
    goal,
)

greedy_result = greedy_best_first_search(
    graph,
    start,
    goal,
    manhattan_distance,
)

astar_result = a_star_search(
    graph,
    start,
    goal,
    manhattan_distance,
)

print("UCS")
print("Path:", ucs_result.path)
print("Cost:", ucs_result.path_cost)
print("Expanded:", ucs_result.nodes_expanded)

print()

print("Greedy")
print("Path:", greedy_result.path)
print("Cost:", greedy_result.path_cost)
print("Expanded:", greedy_result.nodes_expanded)

print()

print("A*")
print("Path:", astar_result.path)
print("Cost:", astar_result.path_cost)
print("Expanded:", astar_result.nodes_expanded)
```

A* with Heuristic Zero is practically UCS.

```python
ucs_result = uniform_cost_search(
    graph,
    start,
    goal,
)

astar_zero_result = a_star_search(
    graph,
    start,
    goal,
    zero_heuristic,
)

print("UCS Path:", ucs_result.path)
print("A* Zero Path:", astar_zero_result.path)

print(
    "UCS Cost:",
    ucs_result.path_cost,
)

print(
    "A* Zero Cost:",
    astar_zero_result.path_cost,
)
```

If the heuristic is consistent, the cost behavior in A* becomes more regular and usually requires less complex reopening when a node is properly expanded.

For a standard grid with four-way movement and uniform costs, Manhattan Distance is usually a natural and consistent heuristic.
