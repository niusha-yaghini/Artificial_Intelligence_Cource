```python
weighted_graph = Graph(directed=False)

weighted_graph.add_edge("A", "B", 2)
weighted_graph.add_edge("A", "C", 5)

weighted_graph.add_edge("B", "D", 7)
weighted_graph.add_edge("B", "C", 1)

weighted_graph.add_edge("C", "D", 2)
```

```python
bfs_result = bfs(
    weighted_graph,
    start="A",
    goal="D",
)

print("Path:", bfs_result.path)
print("Depth:", bfs_result.solution_depth)
print("Cost:", bfs_result.path_cost)
```

Unlike BFS and DFS, in UCS we do not have visited = {start} and instead we have best_cost because if a path with a lower cost is found, we will replace it.

```python
ucs_result = uniform_cost_search(
    weighted_graph,
    start="A",
    goal="D",
)

print("Success:", ucs_result.success)
print("Path:", ucs_result.path)
print("Depth:", ucs_result.solution_depth)
print("Cost:", ucs_result.path_cost)
print("Traversal:", ucs_result.traversal_order)
print("Expanded:", ucs_result.nodes_expanded)
print("Generated:", ucs_result.nodes_generated)
```

```python
bfs_result = bfs(
    weighted_graph,
    "A",
    "D",
)

ucs_result = uniform_cost_search(
    weighted_graph,
    "A",
    "D",
)

print("BFS")
print("Path:", bfs_result.path)
print("Depth:", bfs_result.solution_depth)
print("Cost:", bfs_result.path_cost)

print()

print("UCS")
print("Path:", ucs_result.path)
print("Depth:", ucs_result.solution_depth)
print("Cost:", ucs_result.path_cost)
```

If all edges have equal cost, UCS is the same as BFS.

To ensure standard UCS behavior, the edge costs must not have negative costs. If we have negative edges, other algorithms such as Bellman-Ford come into play.

For now, we consider our search problem with non-negative costs.
